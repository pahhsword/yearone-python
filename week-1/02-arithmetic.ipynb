# Arithmetic within Python

## Basic Arithmetic:
This will look at the starting parts:
- Add (+)
- Subtract (-)
- Divide (/)
- Multiply (*)

```python
print(5+4)
print(34-3)
print(25/5)
print(12*5)
```

## More Advanced:
- Floor (//) - Division: Rounding down -- DIV
- Exponential (**) - To the power off 
- Modulus (%) - Division: Remainder -- MOD

Use cases: Get hours with Floor and get Minutes with Modulous
i.e. 60 // 60 = 1 hour, 60 % 60 = 0 minutes
i.e. 65 // 60 = 1 hour, 65 % 60 = 5 minutes

Exponential can be useful when using powers i.e. PI * (r ** 2) is equivilant to pi * r^2

```python
# Returns the whole numbers that can fit into it
print(26 // 5)
print(29 // 5)
print(29.5 // 5) # Any float will return a number with a float

# Exponential (to a power)
print(2 ** 4) # 2^4 = 16
print(2 ** 3) # 2^3 = 8

print(26 % 5) # Returns the remainder of the division
```
